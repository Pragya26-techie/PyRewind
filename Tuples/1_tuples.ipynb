Inshort,Tuple is immutable list.A tuple can not be changed in any way once created.

#### t1() shows type(*tuple) but t1("pragya") shows type(* str),why?
- because python identifes tuple by the comma,not by the praentheses alone
- In Python, () represents an empty tuple by default. But when we put a single value inside parentheses, like ("bappy"), Python treats it as that original value. We need a comma to create a tuple containing just one element.”

```python
# create a tuple with sigle-item
t = ("pragya",)
t
type(t)

#homo
th = (1,2,3,4)
th

#hetro
t2=(1,2,3,True,[1,2,3])
print(t2)

# nested tuple
t3 = (4,8,9,("a","b","c"))
print(t3)

# using type conversion
t4 = tuple('hello')
print(t4)
```

#### Tuples methoda/functions:
#### len/min/max/sorted/sum

```python
t = (2,5,7,9,13,40,35)
print(min(t))
print(max(t))
print(sum(t))
print(len(t))
print(sorted(t))
print(sorted(t,reverse=True))
```

#### Difference between list and tuple:
- syntax
- mutability
- speed - depends on mutablitablity,fast if it is immutable,slow if its mutable
- memory-takes less space than your list
- built in functionality-list has more
- error prone
- usablity

```python
## checking memory of list and tuple
import sys
L = list(range(1000))
T = tuple(range(1000))

print('List size',sys.getsizeof(L))
print('List size',sys.getsizeof(T))
```

```python
## checking error prone
a = [1,2,3,4]
b = a
a.append(5)
print(a)
print(b)
```

```python
a = (1,2,3)
b = a

a = a + (4,)
print(a)
print(b)
```

#### tuple unpacking


```python
a,b,c = (1,2,3)
print(a,b,c)
```

```python
a,b,c = [1,2,3]
print(a,b,c)
```

```python
a,b= (1,2,3)
```

```python
## we want only few value to be unpacked
a,b,*unwanted = (1,2,3,4,5)
print(a,b)
print(unwanted)
```

```python
## zipping tuples
a = (1,2,3,4)
b = (5,6,7,8)
tuple(zip(a,b))
```
