#### Logging:
"Logging is th eprocess of recording events that occur during a program's execution.we use logging to monitor application behaviour,debug issues,investigate errors,and trac susicious activities.- - It is  especially useful in production environments because it helps developers understand what happened without having to reproduce every issue manually.

#### Exception handling v/s logging
- Exception handling and logging serve different purposes. Exception handling allows us to manage runtime errors gracefully and control how the application responds when something goes wrong.
- Logging records application events, warnings, and error details for debugging, monitoring, and future investigation. We often use them together: exception handling manages the error, while logging preserves useful information about it for developers."
- Logging an error doesn't automatically resolve it, and catching an exception doesn't automatically create a log. That's why combining them is useful in real-world applications.


```python
import logging
import os
```

```python
# create alog folder
LOG_DIR = "logs"
LOG_FILE_NAME = "application.log"

os.makedirs(LOG_DIR,exist_ok = True)
log_path = os.path.join(LOG_DIR,LOG_FILE_NAME)
# print(log_path)
logging.basicConfig(
    filename = log_path,
    format = "[%(asctime)s] %(name)s - %(levelname)s - %(message)s", #format to define log
    level = logging.INFO,   #information of file
    force = True
)
```
