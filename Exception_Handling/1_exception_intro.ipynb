There are 2 stages where error may happen in a program
- During compilation- Syntax error
- During execution - Exceptions

#### Exceptions in Python
An exceprtion ia an event,which occurs during the execution of a program that disrupts the normal flow of the program's instructions.In general,when a Pyhton scripts encounters a situation that it cannot cope with,it raises an exception.An exception is a Python object that represents an error.

When a Python raises an exception.it must either handle the exception immediately otherwise it terminates and quits.

in short- if things go wrong during execution of the program(runtime).it generally happens when something unforseen has happend.

*Examples*
- Memory oveflow
- Divide by zero
- Dtabase error

```python
# Examples of syntax error
print 'hello world'
```

```python
def divide(num):
    r = 12/num
    print(r)
```

```python
divide(0)
```

Why it is important to handle exceptions?
- Exception handling is important because, although a program can run without it, an unexpected runtime error can terminate the program's execution. Exception handling allows us to handle such errors gracefully, provide meaningful error messages, and maintain the normal flow of the application wherever possible. It also helps prevent sensitive technical details from being exposed to users, which can support application security."
- whenever u trying to handle external file or external code u must use try except block preventing fearing error as well as security reason.

##### How to handle exceptions->Try except block

```python
# let's create a file
with open('sample.txt','w') as f:
    f.write('hello world')
```
