```python
def multiply(*args):
    product = 1
    for i in args:
        product = product*i
    return product
```

```python
res = multiply(1,2,3,4,5,6)
print(res)
```

#### **kwargs
**kwargs allow us to pass any number of keyword arguments.
keyword arguments mean that they contain a key-value pair,like a Python dictionary

```python
def display(**kwrags):
    print(kwrags)
```

```python
display(inida="Delhi",Banagladesh="dhaka",srilanka="columbo")
```

```python
def display(**kwargs):
    for (key,val) in kwargs.items():
        print(f"{key} -> {val}")
```

```python
display(inida="Delhi",Banagladesh="dhaka",srilanka="columbo")
```
