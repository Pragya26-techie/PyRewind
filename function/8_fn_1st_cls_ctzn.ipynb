#### Functions are 1st class citizens
it acts like data type
- the operation you perform with other type of data can perform with function as well.
- “Functions are called first-class citizens in Python because Python treats them as objects that can be used like other values. We can assign functions to variables, pass them as arguments, and return them from other functions. The term first-class refers to these capabilities, not to functions being more important than other objects.”

```python
# type and id
def square(num):
    return num**2

print(type(square))
print(id(square))
```

```python
# reassign

x = square
print(x(4))
print((x))

```

```python
# deleting a function
# del square
# print(square)
```

```python
# storing
L = [1,2,3,4,5,square]
L
L[-1](4)
```

```python
# check if function is immutable or mutable object
s = {square}
s
```

* we stored function inside dictionary that stores only immutable object,this shows function is immutable objects

```python
# returning a function

def f():
    def g(a,b):
        return a+b
    return g
val =  f()
print(val)

val = f()(2,3)
print(val)
```

```python
# function as argument

def func_a():
    print("inside func_a")

def func_b(z):
    print('inside func_c')
    return z()

print(func_b(func_a))
```
