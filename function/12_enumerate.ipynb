#### Enumerate function in Python
- Enumerate function will help u a lot whenever u r working with indexes.

```python
L = ["Mayank","Ashu","Ritu","Juhi","Prakhar"]
for i in L:
    print(i)
```

```python
L = ["Mayank","Ashu","Ritu","Juhi","Prakhar"]
for i in enumerate(L):
    print(i)
```

```python
# want to remove tuple from above output format
L = ["Mayank","Ashu","Ritu","Juhi","Prakhar"]
for index,name in enumerate(L):
    print(index,name)
```

```python
# want to stat indexing elements from 1
L = ["Mayank","Ashu","Ritu","Juhi","Prakhar"]
for index,name in enumerate(L,start=1):
    print(index,name)
```

```python
# write some logical program also for specific condition
L = ["Mayank","Ashu","Ritu","Juhi","Prakhar"]
for index,name in enumerate(L,start=1):
    if index == 4:
        print(index,name)
```
