#### Lambda Function
- A lambda function is a small anonymous function.
- A lambda function can take any  number of arguments,but can only have one expression

```python
## square of a function
lambda x:x**2
```

- noraml function return value but lambda return func object lamda

```python
a = lambda x:x**2
a(6)
```

```python
a = lambda x,y:x+y
```

```python
a(4,6)
```

#### Diff between lamda vs Normal Function

- No name
- lamda has no return value(infact returns a function)
- lambda is written in 1 line using *lamda* keyword
- not reusable

Then why use lambda functions?
**They are used with HOF**

```python
# check if a string has 'a'
a = lambda s:'a' in s
a("ritu")
```

```python
a("pargya")
```

```python
# odd and even
eo = lambda x: 'even' if x%2 == 0 else 'odd'
```

```python
eo(9)
```

#### Higher Order Functions
- if any function takes input in another function it is called high-order function

```python
L = [1,2,3,4,5]

# find square of each number
for i in L:
    print(i**2)
```

```python
# need of using hof
L = [1,2,3,4,5]

def square(x):
    return x**2

def transform(f,L):
    output = []
    for i in L:
        output.append(f(i))
    print(output)
```

```python
transform(square,L)
```

```python
transform(lambda x:x**4,L)
```
