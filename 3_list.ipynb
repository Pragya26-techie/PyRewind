#### List Comprehension
List comprehension provides a concise way of creating lists.
* newlist = [expression for item in iterable if condition==True]
* Advantages of List Comprehension
- More time efficient and space efficient than loops.
- Require fewer lines of code.
- Transforms iteratative statement into a formula

```python
# Add 1 to 10 number to a lsit
L = []
for i in range(1,11):
    L.append(i)
print(L)
```

```python
# using List Comprehension
L = [i for i in range(1,11)]
print(L)
```

```python
# sclar multipliaction on a vector
v = [2,3,4]
s = -3
L = []
for i in v:
    result = i * s
    L.append(result)
print(L)
```

```python
L = [i*s for i in v]
print(L)
```

```python
# Add squares
L = [1,2,3,4]
LC = [i**2 for i in L]
LC
```

```python
## print all numbers divisible by 5 in the range of 1 to 50

for i in range(1,51):
    if i%5==0:
        print(i)
```

```python
L = [i for i in range(1,51) if i%5==0]
print(L)
```

```python
# find languages which start with letter p
languages = ['java','python','php','c','javascript']

lc = [lang for lang in languages if lang.startswith('p')]
lc
```

#### Zip
The zip() function returns a zip object, which is an iterator of tuples where the first item in each passed iterator is paired together, and then the second item in each passed iterator are paired together.

If the passed iterators have different lengths, the iterator with the least items decides the length of the new iterator.

```python
L1 = [1, 2, 3, 4]
L2 = [5, 6, 7, 8]
```

```python
list(zip(L2, L1))
```

```python
roll_number = [1,2,3,4,5]
student_name = ['Bappy',"Alex", "Harish", "Kumer","Ahmed"]

list(zip(roll_number, student_name))
```

```python
## subtract operation for two diff list
L1 = [1,2,3,4]
L2 = [5,6,7,8]

[i-j for i,j in zip(L1,L2)]
```

```python
L1 = [1,2,3,4]
L2 = [5,6,7,8]

[i-j for i,j in zip(L2,L1)]
```

```python
L1 = [1, 2, 3, 4]
L2 = [5, 6, 7, 8]
L3 = [9,10]
list(zip(L1,L2,L3))
```
