#### Collection of class objects in OOP

```python
class Person:
    def __init__(self,name,country):
        self.name = name
        self.country = country
```

```python
p1 = Person("Neha","U.S.A")
p2 = Person("Mayank","U.k")
p3 = Person("Tannu","india")
```

```python
L = [p1,p2,p3]
print(L)
```

```python
# now you can apply simple list operation here

for i in L:
    print(i)  
```

```python
for i in L:
    print(i.name,i.country)
```

NOTE- Whatever data structure support mutable data type you can store class object in that data structure
