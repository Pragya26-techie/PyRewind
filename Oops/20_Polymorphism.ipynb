#### Polymorphism 
Polymorphism refers to having multiple forms.Polymorphism is a programming term that refers to the use of same function name,but with different signatures.

- Method Overriding
- Method Overloading
- Operator Overloading

```python
class Shape:
    def area(self,radius):
        return 3.4*radius*2

    def area(self,l,b):
        return l*b

s = Shape()
s.area(2)
```

In above example you can observer even though the first fn name area has one parameter radius stills the code is not executing and throwing error becoz in another language polymorphism is supported but python does not support polymorphism,directly you can't use polymorphism concept inside python

```python
class Shape:
    def area(self,radius,l,b):
        if b == None:
            print("Radius")
            return 3.14*radius**2
        else:
            print("Multiply")
            return l*b
s = Shape()
s.area(radius = 8,l=45,b=4)
```
