```python
class Person:
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender

# outside of class
def greeting(person):
    person.name = "Alex"
    return person
    
```

```python
p = Person("Bappy","Male")
```

```python
print(id(p))
```

```python
a = greeting(p)
```

```python
a.name
```

```python
p.name
```
