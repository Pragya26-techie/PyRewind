No,we cannot access private property of parent class inside child class,we nedd to use getter method

```python
# lets use getter method so that we can access private variable of parent class
class Phone:
  def __init__(self, price, brand, camera):
    print("Inside Phone constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def get_price(self):
     return self.__price


  def buy(self):
    print("Buying a Phone")


class SmartPhone(Phone):
    def print_price(self):
       print(self.get_price())


S  = SmartPhone(200000, "apple", 3)
S.print_price()
```
