```python
class AtmMachine:
    # constructo(special function) ->superpower
    def __init__(self): #data propety
        print(id(self))
        self.pin = ""
        self.balance = 0
        # print("I am executing")
        # self.menu()
    
    def menu(self):#self->locating it is atmmachine function,not any other classs's fn.
        user_input = input(
            """
            Hi how can i help you?
            1. press 1 to create pin
            2.press 2 to change pin
            3. press 3 to check balance
            4. press 4 to withdraw
            5. Anything to exit 
            """
        )

        if user_input == "1": #functional property
            
            self.create_pin()

        elif user_input == "2":
            #change pin
            self.change_pin()

        elif user_input == "3":
            # check balance
            self.check_balance()
        elif user_input == "4":
            # withdraw
            self.withdraw_balance()
        else:
            exit()
    
    def create_pin(self):
        user_pin = input("Enter your pin:")
        self.pin = user_pin

        user_balance = int(input("Enter balance:"))
        self.balance = user_balance
        print("Pin Created Successfully")
        self.menu()

    def change_pin(self):
        old_pin = input("Enter your old pin:")

        if old_pin == self.pin:
            new_pin = input("Enter new pin")
            self.pin = new_pin
            print("Pin changed succesfully!")
            self.menu()
        else:
            print("Invalid Pin")
            self.menu()

    def check_balance(self):
        user_pin = input("Enter your pin:")
        if user_pin == self.pin:
            print(f"your balance is {self.balance}")
        else:
            print("Your pin is incorrect,please try again")
            self.menu()

    def withdraw_balance(self):

        user_pin = input("Enter your pin")
        if user_pin == self.pin:
            #allow to withdraw
            amount = int(input("Enter the amount:"))
            if amount <= self.balance:
                self.balance = self.balance-amount
                print(f"you have withdrawn {amount}.Your new balnce is {self.balance}")
            else:
                print("Insufficient balance")
        else:
            print("Your pin is incorrect,please try again")
        self.menu()
```

In Python Object-Oriented Programming, self is a reference that points to the specific instance (object) of a class currently being used. It allows you to access the attributes and methods belonging to that specific object in memory

__why self important__
- Accessing Instance Properties: By using self, Python knows you are referring to the specific data (like variables) associated with that unique object rather than general class-level data.

- Method Communication: It enables different methods within the same class to call each other. For example, if you want to call a menu() method from inside a create_pin() method, you use self.menu() to ensure the program recognizes it as part of the same object.

- Memory Management: The self keyword confirms that the method or variable belongs to the exact memory location of the object you initialized





- Instance-level data (using self): These are variables unique to a specific object (or instance). For example, if you have an ATM class, the balance variable is instance-level data because one ATM might have $500 while another has $1,000. When you use self.balance, Python accesses the specific balance associated with the object you are currently interacting with.

- Class-level data: This is information shared by all instances of that class. Imagine a variable like bank_name inside the ATM class. Every single ATM object would share the exact same bank_name value. It is defined at the class level, not the instance level, so it doesn't need self to be accessed and remains consistent across every object created from that class.

The Key takeaway:
- Use self to manage data that changes from one object to another (unique state).

- Avoid self if you are defining data that should be identical for every object created from that class (shared state).

__Example__:
class BankAccount:

*Class-level variable (shared by all accounts)*
bank_name = "Global Bank"

def __init__(self, owner, balance):
    # Instance-level variables (specific to each account)
    self.owner = owner
    self.balance = balance

*Creating two different objects (instances)*

account1 = BankAccount("Alice", 500)
account2 = BankAccount("Bob", 1000)

*Accessing class-level data*
print(account1.bank_name)  # Output: Global Bank
print(account2.bank_name)  # Output: Global Bank

*Accessing instance-level data*

print(account1.owner)      # Output: Alice
print(account2.owner)      # Output: Bob
print(account1.balance)    # Output: 500
print(account2.balance)    # Output: 1000

In this code:

- bank_name is a class variable; it is defined directly under the class and stays the same for account1 and account2.

- self.owner and self.balance are instance variables; they are tied to self, meaning each object keeps its own unique value.
