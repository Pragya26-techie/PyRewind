#### Super_Keyword

with "super()" keyword you can access parent class attribute as well as method.

```python
class Phone:
  def __init__(self, price, brand, camera):
    print("Inside Phone constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def get_price(self):
     return self.__price


  def buy(self):
    print("Buying a Phone inside Phone constructor")


class SmartPhone(Phone):
   def buy(self):
    print("Buying a Phone inside SmartPhone constructor")
    super().buy()
    

S  = SmartPhone(200000, "apple", 3)
S.buy()
```
