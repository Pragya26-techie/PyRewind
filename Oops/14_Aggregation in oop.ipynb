- Means you can own different class,but it does not mean you can own private property of that class

for allowing to acess private property of a that class we will use getter and setter method

```python
# here we r taking city as private variable and checking whether customer class could acces it or not
class Customer:
    def __init__(self,name,gender,address):
        self.name = name
        self.gender = gender
        self.address = address

    def print_info(self):
        print(f"Name:{self.name},Gender:{self.gender},Address:{self.address.get_city()},{self.address.pin},{self.address.state}")

class Address:
    def __init__(self,city,pin,state):
        self.__city = city
        self.state = state
        self.pin = pin

    # getter method
    def get_city(self):
        return self.__city
```

```python
add = Address("Bengaluru",56008,"Karnataka")
cus = Customer("Pragya","Female",add)
cus.print_info()
```
