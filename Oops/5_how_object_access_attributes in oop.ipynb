#### How object access attributes in oop

```python
class Person:
    def __init__(self,name,country):
        self.name = name
        self.country = country

    def greeting(self):
        if self.country == "india":
            print(f"Namaste{self.name}")
        
        else:
            print(f"Hello{self.name}")
```

```python
p = Person("pargaya", "USA")
```

```python
p.greeting()
```

```python
p.name
```

```python
p.country
```

```python
p.gender
```

```python
# Attribute creation from outside of the class
p.gender = "male"
```

```python
p.gender
```

It is possible to create attributes outside of class too
