#### Abstraction in oop
Abstraction is a fundamental concept in Python programming that allows to simplify complex cocncepts and focus on the essential details.It invovlves hiding unnecessary details and exposing only the relevant information to the users.

```python
# Abstract Base Class (ABC)
from abc import ABC,abstractmethod

#senior
class BankServer(ABC):
    def database(self):
        print("connected to database")

    @abstractmethod
    def security(self):
        print("Security passed")

#junnior
class MobileApp(BankServer):
    def mobile_login(self):
        print("Mobile app login successful")
```

```python
mob = MobileApp()
```

```python
# Abstract Base Class (ABC)
from abc import ABC,abstractmethod

#senior
class BankServer(ABC):
    def database(self):
        print("connected to database")

    @abstractmethod
    def security(self):
        print("Security passed")

#junnior
class MobileApp(BankServer):
    def mobile_login(self):
        print("Mobile app login successful")
    
    def security(self):
        print("Mobile app security")
```

```python
mob = MobileApp()
```

```python
mob.mobile_login()
```
