#### Reference Variable

```python
class Person:
    def __init__(self):
        self.name = "Pragya"
        self.country = "India"
```

```python
p = Person()
```

```python
id(p)
```

```python
q = p
id(p)
```

```python
print(p.name)
print(q.name)
```

```python
q.name = "yuvan"
```

```python
print(p.name)
print(q.name)
```

Whatever changes you are dong in one object it will reflect to all object,this is called reference variable.
