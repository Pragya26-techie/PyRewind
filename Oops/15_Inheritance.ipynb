#### Inheritance in OOP
Inheritance allows us to define a class that inherits all the methods and properties from another class.Parent class is the class being inherited from,also called bse class.Child class is the class that inherits from another class,also called derived class.
- non private attribute
- non private methods

```python
class User:
    def __init__(self,name,gender):
        self.name = "Pragya"
        self.gender = "Female"

    def login(self):
        print("Login done")

    def register(self):
        print("Register done")


class Student(User):
    def __init__(self):
        self.roll_no = 22

    def enroll(self):
        print("Enrolled the course!")

class Instructor(User):
    def __init__(self):
        self.idno = 22

    def create_course(self):
        print("course created")
```

When a child class defines its own constructor, it overrides the parent class constructor. As a result, the parent's instance variables are not initialized automatically. We use super() to call the parent constructor so that the child object can initialize and access the parent's instance variables along with its own attributes.
