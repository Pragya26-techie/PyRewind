#### Update no of items available

```python
cars_name = ["Audi", "BMW", "TATA", "Volvo"]
no_of_cars = [10,20,30,5]

for i,j in zip(cars_name,no_of_cars):
    print(i,j)
```

You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.

```python
L = [2,4,6,10,1]
result = []

# sum = 0
for i in L:
    s = 0
    for j in L:
        if i <= j:
           s = s + j 
    result.append(s)
print(result)
```

#### Find list of common unique items from two list. and show in increasing order

```python
num1 = [23,45,67,78,89,34,67]
num2 = [34,89,55,56,39,67,67]

result = []

for i in num1:
    for j in num2:
        if i == j and i not in result:
            result.append(i)

result.sort()
print(result)
```
