#### Pickling in Python
- Pickling is the process whereby a python object hierarchy is converted into a byte stream,and */unpickling/* is the inverse operation,whereby a bythe stream (from a binary file or bytes-like object) is convereted back into an object hierarchy.

- use case-in machine learning we save one train model in binary model and one prediction model there we save file as pkl format.

```python
def display_info():
    return "Hi my name is pragya and i m data scientist"
```

```python
d = display_info()
```

```python
# pickle dump
import pickle
with open('function.pkl','wb') as f:
    pickle.dump(d,f)
```

```python
# pickle load
import pickle
with open('function.pkl','rb') as f:
    p = pickle.load(f)
    print(p)
```
