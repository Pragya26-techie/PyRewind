1.write a function which count number of lines and number of words in a text.
- a> read speech_brack.txt find and count number of lines and words.
- b> do same for remaining 3 files

```python
def count_lines_words(filename):
    with open(filename,"r",encoding="utf-8") as file:
        lines = file.readlines()

        num_lines = len(lines)
        num_words = sum(len(line.split()) for line in lines)

        return num_lines,num_words
```

```python
files = [
    "speech_barack_obama.txt",
    "speech_michelle_obama.txt",
    "speech_donald_trump.txt",
    "speech_melina_trump.txt"
]

for filename in files:
    lines, words = count_lines_words(filename)
    print(f"{filename}: {lines} lines, {words} words")
```

2 - Read the countrie_data.json data file and create a function that creates a list of the ten most populated countries.

```python
import json

def most_populated_countries(filename,n):
    with open(filename,"r",encoding="utf-8") as f:
        countries = json.load(f)
    
    sorted_countries = sorted(
        countries,
        key = lambda country: country["population"],
        reverse = True
    )

    result = []
    for country in sorted_countries[:n]:
        result.append({
            "country":country["name"],
            "population":country["population"]
        })
    return result

print(most_populated_countries(filename="countries_data.json", n=10))
```

3 - Raed the countries_data.json data file in data directory,create a function that finds the ten most spoken language

```python
import json

def most_spoken_languages(filename,n):
    with open(filename,"r",encoding = "utf-8") as f:
        countries = json.load(f)

    language_count = {}
    for country in countries:
        for language in country["languages"]:
            if language in language_count:
                language_count[language] += 1
            else:
                language_count[language] = 1
    sorted_languages = sorted(
        language_count.items(),
        key = lambda item: item[1],
        reverse = True
    )

    result = []
    for language,count in sorted_languages[:n]:
        result.append({
            "language":language,
            "count":count
        })
    return result
print(most_spoken_languages("countries_data.json",10))
```
