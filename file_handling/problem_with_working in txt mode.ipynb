#### Problems with working iin text mode
- can't work with files like images
- not good for other data types like int/list/tuples

```python
## working with binary file
with open("iit-kharagpur.png","r") as f:
    print(f.read()) #-> error
```

```python
## working with binary file
with open("iit-kharagpur.png","rb") as f:
    print(f.read())
```

```python
# opening binary file
with open("iit-kharagpur.png",'rb') as f:
    with open("iit-kharagpur_copy.png",'wb') as wf:
        wf.write(f.read())
```

```python
# working with other data types
with open('sample.txt','w') as f:
    f.write(5) #->Error
```

```python
with open('sample1.txt','w') as f:
    f.write("5") 
```

```python
# more complex data
d = {
    'name':'Pragya',
    'age':22,
    'gender':'female'
}
with open('sample1.txt','w') as f:
    f.write(d) #error
```

```python
d = {
    'name':'Pragya',
    'age':22,
    'gender':'female'
}
with open('sample1.txt','w') as f:
    f.write(str(d))
```
