#### File Handling in Python

Types of data used for I/O:
- text:'12345' as a sequence of unicode chars.
- Binary-12345 as a sequence of bytes of its binary equivalent.

Hence there are 2 file types to deal with
- Text files-All program files are text files
- Binary files-Images,music,video,exe files

##### HOW file I/O is done in most programming languages
- open a file
- Read/Write data
- Close the file

```python
## create a file
f = open("sample.txt","w")
f.write("Hello guys")
f.close()

# f.write("Hello")->AFter closing file you wont be able to write
```

```python
# case 2- if the file is already present and u r writing something in write mode the prev txt might be override by new one
f = open('sample.txt','w')
f.write('frustated me')
f.close()
```

- in existing text file if u want to preserve old txt and add new txt insted of 'w' mode open in append mode 'a'

```python
f = open('sample.txt','a')
f.write('\nnot ok')
f.close()
```

#### writelines()

```python
L = ['hello\n','hi\n','how are you\n','I am fine']

f = open('sample.txt','w')
f.writelines(L)
f.close()
```

#### reading from files

```python
f = open('sample.txt','r')
print(f.read())
f.close()
```

```python
# reading upto n chars
f = open('sample.txt','r')
s = f.read(10)
print(s)
f.close()
```

```python
# readline() -> to red line by line
f = open('sample.txt','r')
print(f.readline(),end='')
print(f.readline(),end='')
f.close()
```

```python
# reading entire text using readline
f = open('sample.txt','r')

while True:
    data = f.readline()
    if data == '':
        break
    else:
        print(data,end='') #end->dont need to take xtra space

f.close()
```
