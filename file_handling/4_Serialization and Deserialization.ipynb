- How we can read nd write other type data in file format except txt format here we use concept of derialization and deserialization

#### Serialization an Deserialization
- *Serialization*- Process of converting python data types to JSON format.
- *Deserialization*- process of converting JSON python data types.

##### What is JSON?
The full form of JSON is javascript Object Notation.
- why json format -> because it is supported by all programming language.

```python
# serialization using json module
# list

import json

L = [1,2,3,4,5,6]
with open("test.json",'w') as f:
    json.dump(L,f)
```

```python
# dict
d = {
    'name':'Pragya',
    'age':22,
    'gender':'male'
}
with open('test.json','w') as f:
    json.dump(d,f,indent = 4)
```

```python
# deserialization
import json

with open('test.json','r') as f:
    data = json.load(f)
    print(data)
    print(type(data))
```
