#### File Handling using Context Manager(with)
- it's a good idea to close a file after usage as it will free up the resources
- if we don't close it,garbage collector would close it
- *'with'* keyword closes the file as soon as the usage is over

```python
with open("sample.txt",'w') as f:
    f.write("hello there!")
```

```python
# read()
with open("sample.txt",'r') as f:
    print(f.read())
```

```python
# readline()
with open("sample.txt",'r') as f:
    print(f.readline())
```

```python
# append mode()
with open("sample.txt",'a') as f:
    s = f.write("\nMy name is pargya")
```

```python
# reading only limited char
with open("sample.txt",'r') as f:
    print(f.read(10))
    print(f.read(10))
    

```

- benefits of using read(10) that it will read char line by line and memory will not get crashed

```python
# benefit? -> to load a big file in memory
big_L = ['hello world' for i in range(1000)]
with open('big.txt','w') as f:
    f.writelines(big_L)
```

```python
with open('big.txt','r') as f:
    chunk_size = 10

    while len(f.read(chunk_size)) > 0:
        print(f.read(chunk_size),end="") #end param removes space
        f.read(chunk_size)
```

```python
# seek and tell function
with open('sample.txt','r') as f:
    print(f.read(10))
    print(f.tell())
```

- tell - in which position ur cursor is present currently

```python
with open('sample.txt','r') as f:
    print(f.read(10))
    f.seek(0)
    print(f.tell())
```

```python
with open('sample.txt','w') as f:
    f.write("Hello")
    f.seek(0)
    f.write("Da")
    print(f.tell())
```
