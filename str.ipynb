#### startswith()-
This inbuilt method returns True if string starts with specified prefix.If not,return False.
* string.startswith(prefix,start,end)

```python
msg = "hello welcome"
msg.startswith('hello')
```

```python
msg = "hello welcome"
msg.startswith('hello',5,11)
```

```python
## write a program for finding country of mobile number using prefix
mob_no = input("Enter mob number")
if mob_no.startswith('+91'):
    print("mobile number belongs to india")
elif mob_no.startswith('+1'):
    print('mobile number belongs to america')
elif mob_no.startswith('86'):
    print("mobile number belongs to china")
```

#### Strip() -- removing leading and trailing characters from given string.
* Types-- strip(),lstrip(),rstrip()
* strip() -- removes leading and trailing character
* lstrip()-- removing leading character(space is the default leading characters to remove.,if first word is mismatched it would stop there)
* rstrip() -- removes trailing character

```python
# lstrip()--suppose i asked use to enter name and he tyype name after few space..there we use lstrip() to remove that extra space
name = input("Enter your name:")
print(name.lstrip())
```

#### real-world use case of lstrip()--suppose from url we want only domain not there full address like http://www.

```python
urls = ['http://www.cricbuzz.com','http://ww.youtube.com','http://www.facebook.com']
for url in urls:
    url = url.lstrip('htp:/w.')
    print(url)
```

#### Reverse string--using slicing,while loop.for loop


```python
name = input("enter string:")
print("original string is:",name)
r_name = ""
count = len(name)
while count>0:
    r_name = r_name+name[count-1]
    count=count-1
print('reversed string is:',r_name)
```

```python
name = input("enter string:")
print("original string is:",name)
e_string = ""
for i in name:
    e_string = i+e_string
print("reversed string is:",e_string)
```

```python
#reverse string using slicing:
name = input("enter string")
print("original string is:",name)
print("reverse string ",name[::-1])
```
